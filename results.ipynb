```python
models = ["VGG16", "VGG19", "ResNet50"]

accuracy = [0.8553, 0.8224, 0.8618]

classes = ["bumper_dent","bumper_scratch","door_dent","door_scratch",
           "glass_shatter","head_lamp","tail_lamp","unknown"]

precision = {
    "VGG16": [0.83,1.00,0.72,0.75,1.00,0.67,0.85,0.98],
    "VGG19": [0.69,1.00,0.65,0.63,1.00,0.86,0.86,0.96],
    "ResNet50": [0.83,0.50,0.89,0.67,0.92,0.87,0.87,0.96]
}

recall = {
    "VGG16": [0.77,0.50,0.90,0.75,0.85,0.92,0.79,0.95],
    "VGG19": [0.85,0.12,0.85,0.75,0.92,0.92,0.86,0.87],
    "ResNet50": [0.77,0.50,0.80,0.75,0.85,1.00,0.93,0.95]
}

f1 = {
    "VGG16": [0.80,0.67,0.80,0.75,0.92,0.77,0.81,0.96],
    "VGG19": [0.76,0.22,0.74,0.69,0.96,0.89,0.86,0.91],
    "ResNet50": [0.80,0.50,0.84,0.71,0.88,0.93,0.90,0.95]
}
```

```python
import matplotlib.pyplot as plt

plt.figure()
plt.bar(models, accuracy)
plt.xlabel("Techniques")
plt.ylabel("Accuracy")
plt.title("Accuracy Comparison")
plt.ylim(0.75, 0.90)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.arange(len(models))
width = 0.1

plt.figure(figsize=(10,5))

for i, cls in enumerate(classes):
    values = [precision[m][i] for m in models]
    plt.bar(x + i*width, values, width, label=cls)

plt.xticks(x + width*3, models)
plt.xlabel("Techniques")
plt.ylabel("Precision")
plt.title("Precision Comparison")

plt.legend(loc='upper left', bbox_to_anchor=(1,1))
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,5))

for i, cls in enumerate(classes):
    values = [recall[m][i] for m in models]
    plt.bar(x + i*width, values, width, label=cls)

plt.xticks(x + width*3, models)
plt.xlabel("Techniques")
plt.ylabel("Recall")
plt.title("Recall Comparison")

plt.legend(loc='upper left', bbox_to_anchor=(1,1))
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))

for i, cls in enumerate(classes):
    values = [f1[m][i] for m in models]
    plt.bar(x + i*width, values, width, label=cls)

plt.xticks(x + width*3, models)
plt.xlabel("Techniques")
plt.ylabel("F1 Score")
plt.title("F1 Score Comparison")

plt.legend(loc='upper left', bbox_to_anchor=(1,1))
plt.tight_layout()
plt.show()
```
